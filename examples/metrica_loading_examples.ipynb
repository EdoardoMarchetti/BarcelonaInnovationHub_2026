# Metrica - Data Loading Examples

This notebook shows how to load Metrica tracking files using Kloppy, both from local files and remote URLs.

```python
from kloppy import metrica
import pandas as pd
```

## 1. Loading from Local CSV Files

Load Metrica tracking files from local CSV files.

```python
# Load local CSV files
dataset_local = metrica.load_tracking_csv(
    home_data="../data/metrica/Sample_Game_1/Sample_Game_1_RawTrackingData_Home_Team.csv",
    away_data="../data/metrica/Sample_Game_1/Sample_Game_1_RawTrackingData_Away_Team.csv",
    
    # Optional arguments
    sample_rate=1/5,  # Sample every 5 frames (0.2 Hz)
    limit=100,        # Limit to 100 frames for example
    coordinates="metrica"  # Metrica coordinate system (0-1)
)

# Convert to DataFrame for visualization
df_local = dataset_local.to_df()
print(f"Dataset shape: {df_local.shape}")
df_local.head()
```

## 2. Loading from Remote CSV URLs

Load Metrica tracking files directly from remote URLs (e.g., GitHub).

```python
# Load CSV files from remote URLs
dataset_url = metrica.load_tracking_csv(
    home_data="https://raw.githubusercontent.com/metrica-sports/sample-data/master/data/Sample_Game_1/Sample_Game_1_RawTrackingData_Home_Team.csv",
    away_data="https://raw.githubusercontent.com/metrica-sports/sample-data/master/data/Sample_Game_1/Sample_Game_1_RawTrackingData_Away_Team.csv",
    
    # Optional arguments
    sample_rate=1/5,
    limit=100,
    coordinates="metrica"
)

# Convert to DataFrame for visualization
df_url = dataset_url.to_df()
print(f"Dataset shape: {df_url.shape}")
df_url.head()
```

## 3. Loading EPTS Files from Local

Load Metrica tracking files in EPTS format (XML + TXT) from local files.

```python
# Load local EPTS files (if available)
dataset_epts = metrica.load_tracking_epts(
    meta_data="../data/metrica/Sample_Game_3/Sample_Game_3_metadata.xml",
    raw_data="../data/metrica/Sample_Game_3/Sample_Game_3_tracking.txt",
    
    # Optional arguments
    sample_rate=1/5,
    limit=100,
    coordinates="metrica"
)

# Convert to DataFrame for visualization
df_epts = dataset_epts.to_df()
print(f"Dataset shape: {df_epts.shape}")
df_epts.head()
```

## 4. Full Loading Without Limits

Example of full loading without frame limits, useful for complete analysis.

```python
# Load the full dataset without limits
dataset_full = metrica.load_tracking_csv(
    home_data="../data/metrica/Sample_Game_1/Sample_Game_1_RawTrackingData_Home_Team.csv",
    away_data="../data/metrica/Sample_Game_1/Sample_Game_1_RawTrackingData_Away_Team.csv",
    coordinates="metrica"
    # No limit, no sample_rate = load everything
)

df_full = dataset_full.to_df()
print(f"Full dataset shape: {df_full.shape}")
print(f"\nAvailable periods: {df_full['period_id'].unique()}")
print(f"\nTimestamp range: {df_full['timestamp'].min():.2f} - {df_full['timestamp'].max():.2f} seconds")
```

## Notes

- **sample_rate**: Reduces the sampling frequency (e.g., 1/5 = every 5 frames)
- **limit**: Limits the number of frames loaded (useful for quick tests)
- **coordinates**: Coordinate system ("metrica" = 0-1, "kloppy" = 0-105x68 meters)
- Metrica CSV files contain columns for each player (home_11_x, home_11_y, away_22_x, etc.)
- EPTS files are an alternative format with separate XML metadata
