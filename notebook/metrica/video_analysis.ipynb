```python
print("Start importing libraries")
from transformers import AutoProcessor, AutoModelForImageTextToText
import torch
print("Importing libraries done")
```

```python
model_path = "HuggingFaceTB/SmolVLM2-256M-Video-Instruct"  # or whatever the actual model name is
processor = AutoProcessor.from_pretrained(model_path)


device = "cuda" if torch.cuda.is_available() else "cpu"
model_kwargs = {
    "torch_dtype": torch.bfloat16 if device == "cuda" else torch.float32,
}
if device == "cuda":
    model_kwargs["_attn_implementation"] = "flash_attention_2"

model = AutoModelForImageTextToText.from_pretrained(
    model_path,
    **model_kwargs
).to(device)
```

```python
import os

# Oppure se il file è nella stessa directory del notebook:
video_path = "tracking_animation.mp4"  # senza il doppio .mp4

messages = [
    
    {
        "role": "user",
        "content": [
            {"type": "video", "path": video_path},  # Corretto: solo .mp4
            {
                "type": "text",
                "contet":(
            "Eres un analista táctico de fútbol profesional. Analiza la siguiente secuencia de eventos de una jugada y genera una descripción táctica detallada en español.\n"
            "\n"
            "=== ATTACKING DIRECTION ===\n"
            "Away is attacking from left to right\n"
            "\n"
            "Instrucciones:\n"
            "- Sé conciso en la descripción.\n"
            "- Describe el desarrollo de la jugada de forma narrativa y fluida.\n"
            "- Identifica patrones tácticos (pases cortos, cambios de orientación, progresión, etc.).\n"
            "- Menciona las zonas del campo donde ocurren las acciones principales.\n"
            "- Usa terminología táctica apropiada (ej: \"construcción desde atrás\", \"transición\", \"finalización\").\n"
            "- Sé específico sobre los jugadores y sus acciones cuando sea relevante.\n"
            "- No añadas suposiciones sobre roles de jugadores "
        )
            }
        ]
    },
]


inputs = processor.apply_chat_template(
    messages,
    add_generation_prompt=True,
    tokenize=True,
    return_dict=True,
    return_tensors="pt",
).to(model.device, dtype=torch.bfloat16)

generated_ids = model.generate(**inputs, do_sample=False, max_new_tokens=1000)
generated_texts = processor.batch_decode(
    generated_ids,
    skip_special_tokens=True,
)
print("\n\n\n\n")
print(generated_texts[0])
```
