```python
def load_tracking_as_mplsoccer(home_file, away_file, home_team='Home', away_team='Away'):

    # Cargar los datos
    df_home = pd.read_csv(home_file, skiprows=2)
    df_home.sort_values('Time [s]', inplace=True)

    df_away = pd.read_csv(away_file, skiprows=2)
    df_away.sort_values('Time [s]', inplace=True)

    for df in [df_home, df_away]:
        cols = list(np.repeat(df.columns[3::2], 2))
        cols = [col+'_x' if i % 2 == 0 else col+'_y' for i, col in enumerate(cols)]
        cols = np.concatenate([df.columns[:3], cols])
        df.columns = cols

    df_ball = df_away[['Period', 'Frame', 'Time [s]', 'Ball_x', 'Ball_y']].copy()
    df_home.drop(['Ball_x', 'Ball_y'], axis=1, inplace=True)
    df_away.drop(['Ball_x', 'Ball_y'], axis=1, inplace=True)
    df_ball.rename({'Ball_x': 'x', 'Ball_y': 'y'}, axis=1, inplace=True)


    #Detect attacking direction
    first_home_frame = df_home.loc[0, [c for c in df_home.columns if 'x' in c]]
    first_away_frame = df_away.loc[0, [c for c in df_away.columns if 'x' in c]]

    home_x_mean = first_home_frame.mean()
    away_x_mean = first_away_frame.mean()

    if home_x_mean < away_x_mean: # TODO: siempre va a ser así, en cualquier caso?
        directions = {
            home_team: {1: 1, 2:-1},
            away_team: {1: -1, 2: 1}
        }
    else:
        directions = {
            home_team: {1: -1, 2: 1},
            away_team: {1: 1, 2: -1}
        }
    return df_home, df_away, df_ball, directions
```

```python
def add_direction(df, direction_map):
    df = df.copy()
    df['Attacking_Direction'] = df.apply(
        lambda row: direction_map.get(row['Team'], {}).get(row['Period']), 
        axis=1
    )
    return df

def invert_y_axis(df):
    df = df.copy()
    df['Start Y'] = 1 - df['Start Y']
    df['End Y'] = 1 - df['End Y']
    return df


def invert_y_axis_tracking(df, ball=False):
    df = df.copy()
    if ball:
        y_cols = ['y']
    else:
        y_cols = [c for c in df.columns if c.endswith('_y')]
    df[y_cols] = 1 - df[y_cols]
    return df


def to_coordinates(df, pitch_length=105, pitch_width=68):
    df = df.copy()
    df['Start X'] = df['Start X'] * pitch_length
    df['End X'] = df['End X'] * pitch_length
    df['Start Y'] = df['Start Y'] * pitch_width
    df['End Y'] = df['End Y'] * pitch_width
    return df

def to_coordinates_tracking(df, pitch_length=105, pitch_width=68, ball=False):
    df = df.copy()
    if ball:
        x_cols = ['x']
        y_cols = ['y']
    else:
        x_cols = [c for c in df.columns if c.endswith('_x')]
        y_cols = [c for c in df.columns if c.endswith('_y')]
    df[x_cols] = df[x_cols] * pitch_length
    df[y_cols] = df[y_cols] * pitch_width
    return df
```

```python

def get_event_zones_vectorized(df, pitch_length=1.0, pitch_width=1.0):
    """
    Versión DEFINITIVA: 
    1. Maneja NaNs correctamente (Row 0 será 'Unknown').
    2. Usa tus nombres de columnas (Start_third, Start_channel).
    3. Aplica lógica táctica profesional (Half-spaces estrechos).
    """
    # Trabajamos sobre copia para seguridad
    df = df.copy()
    
    # 1. Asegurar numéricos
    cols = ['Start X', 'Start Y', 'End X', 'End Y']
    for col in cols:
        df[col] = pd.to_numeric(df[col], errors='coerce')

    def compute_spatial_data(prefix):
        x = df[f'{prefix} X']
        y = df[f'{prefix} Y']
        dir_atk = df['Attacking_Direction']
        
        # MÁSCARA DE VALIDEZ: True si tenemos coordenadas, False si es NaN
        valid = x.notna() & y.notna()
        
        # Arrays para resultados, llenos de 'Unknown' por defecto
        res_third = np.full(len(df), 'Unknown', dtype=object)
        res_channel = np.full(len(df), 'Unknown', dtype=object)
        res_zone_num = np.full(len(df), np.nan)
        
        # --- CÁLCULO SOLO CON DATOS VÁLIDOS ---
        # Extraemos valores limpios para no romper las mates con NaNs
        if valid.any(): # Solo si hay al menos una fila válida
            x_v = x[valid].values
            y_v = y[valid].values
            d_v = dir_atk[valid].values
            
            # Normalización Táctica
            norm_x = np.where(d_v == 1, x_v / pitch_length, 1 - (x_v / pitch_length))
            norm_y = np.where(d_v == 1, y_v / pitch_width, 1 - (y_v / pitch_width))
            
            # 1. Tercios
            thirds = np.select(
                [norm_x <= 1/3, norm_x <= 2/3],
                ['Tercio Defensivo', 'Tercio Medio'],
                default='Tercio Ofensivo'
            )
            
            # 2. Carriles (Standard Táctico: 21-37-63-79)
            channels = np.select(
                [norm_y <= 0.21, norm_y <= 0.37, norm_y <= 0.63, norm_y <= 0.79],
                ['Banda Izquierda', 'Carril Izquierdo', 'Centro', 'Carril Derecho'],
                default='Banda Derecha'
            )
            
            # 3. Zona Numérica (Grid 3x6)
            col_idx = np.clip((norm_x // (1/6)).astype(int), 0, 5)
            row_idx = np.clip((norm_y // (1/3)).astype(int), 0, 2)
            zone_nums = (3 - row_idx) + (3 * col_idx) # Fórmula mágica: (3-row) + 3*col
            
            # Asignamos resultados usando la máscara
            res_third[valid] = thirds
            res_channel[valid] = channels
            res_zone_num[valid] = zone_nums

        return res_third, res_channel, res_zone_num

    # Ejecución Vectorizada
    df['Start_third'], df['Start_channel'], df['Start_Zone_Num'] = compute_spatial_data('Start')
    df['End_third'], df['End_channel'], df['End_Zone_Num'] = compute_spatial_data('End')

    return df
```

```python
def build_pro_tactical_phases_rag_ready(events_df):
    """
    Motor de Identificación de Fases Tácticas para RAG.
    Maneja lógica de posesión avanzada, transiciones de zonas y 
    genera descripciones narrativas para modelos de lenguaje.
    """
    # 1. Preparación de datos
    df = events_df.copy().reset_index() # Reseteamos para tener el ID original del evento
    
    # ---------------------------------------------------------
    # 2. LÓGICA DE POSESIÓN (Algoritmo de Estado Táctico)
    # ---------------------------------------------------------
    possession_owners = []
    current_owner = df.iloc[0]['Team'] if pd.notna(df.iloc[0]['Team']) else 'Unknown'
    
    # Mapeo de comportamiento de eventos
    possession_confirmers = {'PASS', 'SHOT', 'DRIBBLE', 'TOQUE', 'CARRY'}
    set_pieces = {'SET PIECE'}
    interruptions = {'FAULT RECEIVED', 'CARD', 'BALL OUT', 'CHALLENGE', 'FOCUL'}
    turnovers = {'RECOVERY', 'INTERCEPTION'}

    # Iteración rápida por tuplas para el motor de estados
    for row in df.itertuples():
        evt_type = str(row.Type).upper()
        evt_team = row.Team
        
        # Prioridad 1: Balón Parado (Cambio de dueño inmediato y reseteo de fase)
        if evt_type in set_pieces:
            current_owner = evt_team
        # Prioridad 2: Recuperaciones (Turnover directo)
        elif evt_type in turnovers:
            current_owner = evt_team
        # Prioridad 3: Acciones de Posesión (Confirmación)
        elif evt_type in possession_confirmers:
            current_owner = evt_team
        # Prioridad 4: Interrupciones (Mantenemos el dueño actual sin cambios)
        elif evt_type in interruptions:
            pass 
        # Seguridad: Si hay equipo pero evento desconocido
        elif pd.notna(evt_team):
            current_owner = evt_team
            
        possession_owners.append(current_owner)
    
    df['Smart_Owner'] = possession_owners
    
    # Identificar cambio de fase (Cambio de dueño O Saque de centro/banda/etc)
    df['New_Phase'] = (df['Smart_Owner'] != df['Smart_Owner'].shift()) | (df['Type'] == 'SET PIECE')
    df['Phase_ID'] = df['New_Phase'].cumsum()
    
    # ---------------------------------------------------------
    # 3. FUNCIONES DE TRANSICIÓN (Para el RAG)
    # ---------------------------------------------------------
    def get_path(series):
        """Genera la ruta táctica: 'Zona 1 -> Zona 4 -> Zona 6'"""
        items = [str(x) for x in series if pd.notna(x) and str(x).lower() != 'unknown']
        if not items: return "Unknown"
        # Eliminar duplicados consecutivos para limpiar la ruta
        clean_path = [items[0]]
        for i in range(1, len(items)):
            if items[i] != items[i-1]:
                clean_path.append(items[i])
        return " -> ".join(clean_path)

    # ---------------------------------------------------------
    # 4. AGREGACIÓN DE FASE (La Gran Base de Datos)
    # ---------------------------------------------------------
    phases = df.groupby('Phase_ID').agg(
        Start_Frame=('Start Frame', 'min'),
        End_Frame=('End Frame', 'max'),
        # Identificadores
        Team=('Smart_Owner', 'first'),
        Period=('Period', 'first'),
        Start_Time=('Start Time [s]', 'min'),
        End_Time=('End Time [s]', 'max'),
        Duration=('Start Time [s]', lambda x: x.max() - x.min()),
        Event_Count=('Type', 'count'),
        
        # Coordenadas Start/End
        Start_X=('Start X', 'first'),
        Start_Y=('Start Y', 'first'),
        End_X=('End X', 'last'),
        End_Y=('End Y', 'last'),
        
        # TRANSICIONES (KPIs para el Chatbot)
        Zone_Start=('Start_Zone_Num', 'first'),
        Zone_End=('End_Zone_Num', 'last'),
        Zone_Path=('Start_Zone_Num', get_path),
        
        Third_Start=('Start_third', 'first'),
        Third_End=('End_third', 'last'),
        Third_Path=('Start_third', get_path),
        
        Channel_Start=('Start_channel', 'first'),
        Channel_End=('End_channel', 'last'),
        Channel_Path=('Start_channel', get_path),
        
        # Listado detallado para el modelo táctico
        Events_List=('Type', list),
        Subtypes_List=('Subtype', list),
        Events_Ids=('index', list), # Permite cruzar con tracking
        Start_Type=('Type', 'first'),
        End_Type=('Type', 'last')
    ).reset_index()

    # ---------------------------------------------------------
    # 5. RESULTADO Y NARRATIVA
    # ---------------------------------------------------------
    def define_outcome(row):
        evs = row['Events_List']
        subs = [str(s) for s in row['Subtypes_List']]
        last_ev = row['End_Type']
        
        if 'SHOT' in evs:
            if any('GOAL' in s for s in subs): return 'GOL'
            if any('BLOCKED' in s for s in subs): return 'Tiro Bloqueado'
            return 'Tiro'
        if last_ev == 'BALL OUT': return 'Balón Fuera'
        if last_ev == 'BALL LOST': return 'Pérdida de Posesión'
        if last_ev == 'FAULT RECEIVED': return 'Falta Provocada'
        return 'Posesión Finalizada'

    phases['Outcome'] = phases.apply(define_outcome, axis=1)

    # ---------------------------------------------------------
    # 6. CLASIFICACIÓN TÁCTICA Y LIMPIEZA RAG (NUEVO)
    # ---------------------------------------------------------
    
    def classify_phase(row):
        duration = row['Duration']
        n_events = row['Event_Count']
        outcome = row['Outcome']
        
        # Micro-acciones (Despejes, rebotes, acciones defensivas puras)
        if duration < 2.0 and n_events <= 2:
            if outcome in ['Balón Fuera', 'Pérdida de Posesión']:
                return 'Acción Defensiva/Despeje'
            return 'Disputa Breve'
            
        # Posesiones Reales
        if n_events >= 3 or duration > 5.0:
            return 'Posesión Consolidada'
            
        return 'Transición Rápida'

    phases['Phase_Type'] = phases.apply(classify_phase, axis=1)

    # GENERADOR DE TEXTO "HUMANO" PARA EL LLM
    def clean_rag_text(row):
        # Limpiar "Unknown" para que el texto sea elegante
        start_loc = f"{row['Third_Start']} / {row['Channel_Start']}"
        start_loc = start_loc.replace('Unknown', 'Zona No Definida')
        
        path = row['Channel_Path']
        if path == 'Unknown': path = 'Sin desplazamiento significativo'
        
        # Estructura Narrativa para embeddings
        return (
            f"{row['Phase_Type']} de {row['Team']} ({row['Duration']:.1f}s). "
            f"Inicio: {start_loc}. "
            f"Recorrido: {path}. "
            f"Final: {row['Outcome']} en {row['Third_End']}."
        )
    


    # El "Token" para el RAG: Esta frase es lo que el Chatbot entenderá mejor
    # phases['RAG_Description'] = (
    #     "Fase de " + phases['Team'] + " (" + phases['Duration'].round(1).astype(str) + "s). " +
    #     "Origen: " + phases['Third_Start'] + " / " + phases['Channel_Start'] + ". " +
    #     "Ruta: " + phases['Zone_Path'] + ". " +
    #     "Final: " + phases['Outcome'] + " en " + phases['Third_End']
    # )
    phases['RAG_Description'] = phases.apply(clean_rag_text, axis=1)

    return phases
```

```python
epv_filename = './epv_data.csv'
if not os.path.exists(epv_filename):
    print("⚠️ Aviso: No se encontró 'epv_data.csv'. Creando uno por defecto...")
    # Grid genérico simplificado 32x50
    dummy_grid = np.linspace(0.005, 0.05, 50) * np.ones((32, 50))
    pd.DataFrame(dummy_grid).to_csv(epv_filename, header=False, index=False)

epv_grid = pd.read_csv(epv_filename, header=None).values

def get_epv(x, y, grid):
    """Obtiene el valor EPV de una coordenada específica"""
    rows, cols = grid.shape
    c = int(x * cols); c = min(max(c, 0), cols-1)
    r = int(y * rows); r = min(max(r, 0), rows-1)
    return grid[r, c]


```

```python

def build_llm_context(phase_row, events_df, epv_grid, pitch_length=105, pitch_width=68):
    """
    Versión 'Barça Elite': 
    - Reconstrucción de narrativa sin 'Unknowns'.
    - Lógica de progreso y EPV corregida.
    - Tiempo relativo con precisión de milisegundos.
    """
    
    def safe_fmt(val, decimals=2):
        if pd.isna(val) or val is None: return None
        try:
            f_val = float(val)
            return round(f_val, decimals) if not f_val.is_integer() else int(f_val)
        except: return str(val)

    # 1. ACCESO SEGURO A EVENTOS (Usamos diccionarios para evitar errores de índice)
    event_ids = phase_row['Events_Ids']
    local_events = events_df.loc[event_ids].copy()
    events_dicts = local_events.to_dict(orient='records')
    
    # 2. IDENTIFICAR PUNTOS DE ANCLAJE (Evitar nulos del Kick-off)
    # Buscamos el primer y último evento con coordenadas válidas
    valid_events = [e for e in events_dicts if pd.notna(e.get('Start X'))]
    
    if not valid_events:
        return None # O manejar fases sin coordenadas

    first_valid = valid_events[0]
    last_valid = valid_events[-1]
    
    # Datos para el Summary
    team = phase_row['Team']
    direction = first_valid.get('Attacking_Direction', 1)
    
    # 3. CÁLCULO DE MÉTRICAS (EPV e Impacto)
    sx, sy = first_valid['Start X'], first_valid['Start Y']
    ex, ey = last_valid['End X'], last_valid['End Y']
    
    # Progreso Real (Dirección de ataque: 1 -> Der, -1 -> Izq)
    progreso_m = (ex - sx) * pitch_length * direction
    
    # EPV
    try:
        epv_start = get_epv(sx, sy, epv_grid)
        epv_end = get_epv(ex, ey, epv_grid)
        epv_impact = epv_end - epv_start
    except:
        epv_impact = 0.0

    # 4. RECONSTRUCCIÓN DE NARRATIVA (Eliminamos el 'Unknown' de raíz)
    setup_str = f"{first_valid['Start_third']} ({first_valid['Start_channel']})"
    path_str = phase_row['Channel_Path'].replace('Unknown -> ', '').replace(' -> Unknown', '')
    if path_str == 'Unknown': path_str = first_valid['Start_channel']

    narrative = (
        f"{phase_row['Phase_Type']} de {team} ({phase_row['Duration']:.1f}s). "
        f"Inicio: {setup_str}. Ruta: {path_str}. Final: {phase_row['Outcome']}."
    )

    # 5. CONSTRUCCIÓN DEL JSON
    summary = {
        'phase_id': int(phase_row['Phase_ID']),
        'team': team,
        'duration': safe_fmt(phase_row['Duration'], 1),
        'type': phase_row['Phase_Type'],
        'outcome': phase_row['Outcome'],
        'narrative': narrative,
        'tactical': {
            'setup': setup_str,
            'path_zones': phase_row['Zone_Path'],
            'progress_m': safe_fmt(progreso_m, 1),
            'epv_impact': safe_fmt(epv_impact, 4)
        }
    }

    # 6. EVENTOS CON TIEMPO RELATIVO PRECISO
    formatted_events = []
    t_start_phase = events_dicts[0]['Start Time [s]']

    for idx, evt in enumerate(events_dicts):
        # t relativo al inicio de la fase
        t_rel = float(evt['Start Time [s]']) - float(t_start_phase)
        
        # Acción limpia
        action = evt['Type']
        sub = evt.get('Subtype')
        full_action = f"{action} ({sub})" if pd.notna(sub) else action

        formatted_events.append({
          'seq': idx + 1,
          'player': evt.get('From'),
          'action': full_action,
          'start': {
              'x': safe_fmt(evt.get('Start X')), 
              'y': safe_fmt(evt.get('Start Y')), 
              'z': safe_fmt(evt.get('Start_Zone_Num'), 0)
          },
          'end': {
              'x': safe_fmt(evt.get('End X')), 
              'y': safe_fmt(evt.get('End Y')), 
              'z': safe_fmt(evt.get('End_Zone_Num'), 0)
          },
          't': safe_fmt(t_rel, 2) # 2 decimales para capturar pases rápidos
        })

    return {'summary': summary, 'events': formatted_events}
```

```python
import json
import os

# Usamos el router pero con un modelo certificado como "Serverless"
client = OpenAI(
    base_url="https://router.huggingface.co/v1",
    api_key=os.getenv("HF_TOKEN")
)

def generate_with_api(prompt, max_new_tokens=600, temperature=0.2):
    """
    Motor de generación estable.
    Usa Llama 3.1 8B, que es el estándar de oro para evitar errores 403 en el Router.
    """
    try:
        completion = client.chat.completions.create(
            model="meta-llama/Llama-3.1-8B-Instruct", 
            messages=[
                {
                    "role": "system", 
                    "content": "Eres el Analista Táctico del Barça. Sé breve, técnico (interiores, tercer hombre, fijar rivales) y no inventes formaciones si no las ves."
                },
                {"role": "user", "content": prompt}
            ],
            max_tokens=max_new_tokens,
            temperature=temperature
        )
        return completion.choices[0].message.content.strip()
    except Exception as e:
        return f"ERROR_API: {str(e)}"
```

```python
def generate_tactical_insight(phase_row, events_df, epv_grid):
    """
    Genera un análisis táctico profesional basado en conceptos de fútbol 
    de alto rendimiento (Juego de Posición, Tercer Hombre, etc.)
    """
    
    # 1. Obtenemos el contexto enriquecido (el JSON que creamos antes)
    rich_context = build_llm_context(phase_row, events_df, epv_grid)
    
    # 2. PROMPT TÁCTICO UNIVERSAL (Nivel Pro)
    prompt = f"""Actúa como un Analista Táctico de Élite. Tu objetivo es transformar datos en conocimiento estratégico.

=== CONTEXTO DE LA POSESIÓN ===
- Equipo: {rich_context['summary']['team']}
- Tipo: {rich_context['summary']['type']}
- Duración: {rich_context['summary']['duration']}s
- Descripción: {rich_context['summary']['narrative']}
- Métricas: 
    * Avance: {rich_context['summary']['tactical']['progress_m']} metros
    * Impacto EPV: {rich_context['summary']['tactical']['epv_impact']}

=== SECUENCIA DE EVENTOS ===
{json.dumps(rich_context['events'], ensure_ascii=False, indent=2)}

DIRECTRICES DE ANÁLISIS:
1. LENGUAJE PRO: Usa conceptos como 'fijar oponentes', 'hombre libre', 'tercer hombre', 'basculación', 'superioridad posicional' y 'ruptura de líneas'.
2. DINÁMICA: ¿Fue una posesión para atraer presión y saltar líneas o una circulación en U sin profundidad?
3. CLAVE DEL ÉXITO/ERROR: Identifica la acción que cambió el valor EPV. ¿Hubo un pase vertical que rompió el bloque o una pérdida en zona de seguridad?
4. RESULTADO: Analiza si el desenlace ({rich_context['summary']['outcome']}) fue consecuencia de la estructura previa.

Escribe un párrafo técnico, directo y sin introducciones. Enfócate en el 'POR QUÉ' sucedió lo que los datos muestran.
Análisis Técnico:"""

    # 3. Llamada a la API (Temperatura baja para evitar invenciones)
    insight = generate_with_api(
        prompt,
        max_new_tokens=400,
        temperature=0.3 
    )
    
    return insight.strip()
```

```python


def to_long_form(df):
    """ Pivota los datos de tracking de formato ancho a largo """
    df = pd.melt(df, id_vars=df.columns[:3], value_vars=df.columns[3:], var_name='player')
    df.loc[df.player.str.contains('_x'), 'coordinate'] = 'x'
    df.loc[df.player.str.contains('_y'), 'coordinate'] = 'y'
    df = df.dropna(axis=0, how='any')
    df['player'] = df.player.str[6:-2]
    df = (df.set_index(['Period', 'Frame', 'Time [s]', 'player', 'coordinate'])['value']
          .unstack()
          .reset_index()
          .rename_axis(None, axis=1))
    return df

def plot_pro_tracking_dark(
    df_home, 
    df_away, 
    df_ball, 
    start_frame, 
    end_frame, 
    show_names=True, 
    home_team_name="Home", 
    away_team_name="Away", 
    home_color="#ff4b4b", # Rojo Pro
    away_color="#4b88ff", # Azul Pro
    pitch_color="#2b2b2b" # Fondo Oscuro
):
    # 1. SLICING Y PREPARACIÓN
    df_home = df_home.loc[(df_home.Frame >= start_frame) & (df_home.Frame <= end_frame)].copy()
    df_away = df_away.loc[(df_away.Frame >= start_frame) & (df_away.Frame <= end_frame)].copy()    
    df_ball = df_ball.loc[(df_ball.Frame >= start_frame) & (df_ball.Frame <= end_frame)].copy()

    df_home_long = to_long_form(df_home)
    df_away_long = to_long_form(df_away)

    home_players = df_home_long['player'].unique()
    away_players = df_away_long['player'].unique()

    # 2. CONFIGURACIÓN DEL PITCH ESTILO DARK
    pitch = Pitch(pitch_type='uefa', pitch_color=pitch_color, line_color='white', goal_type='line')
    fig, ax = pitch.draw(figsize=(14, 9))
    fig.set_facecolor(pitch_color) # Fondo de la figura completo

    # Objetos animados
    ball_plot, = ax.plot([], [], ms=8, marker='o', color='#ffff00', markeredgecolor='black', zorder=5)
    home_plot, = ax.plot([], [], ms=12, marker='o', color=home_color, markeredgecolor='white', ls='None', zorder=4)
    away_plot, = ax.plot([], [], ms=12, marker='o', color=away_color, markeredgecolor='white', ls='None', zorder=4)

    # 3. CREACIÓN DE ETIQUETAS (Nombres/Dorsales)
    home_texts = []
    away_texts = []
    
    if show_names:
        # Estilo de la caja: fondo semitransparente para que no tape el campo pero sea legible
        bbox_props = dict(boxstyle="round,pad=0.2", fc="white", ec="none", alpha=0.7)
        
        for _ in home_players:
            txt = ax.text(0, 0, "", fontsize=9, color=home_color, ha='center', fontweight='bold', 
                          zorder=10, bbox=bbox_props, visible=False)
            home_texts.append(txt)
        for _ in away_players:
            txt = ax.text(0, 0, "", fontsize=9, color=away_color, ha='center', fontweight='bold', 
                          zorder=10, bbox=bbox_props, visible=False)
            away_texts.append(txt)

    # Título elegante con highlight_text
    fig_text(
        0.5, 0.95, f"<{home_team_name}> vs <{away_team_name}>", 
        size=22, fig=fig, highlight_textprops=[{"color": home_color}, {"color": away_color}],
        ha="center", color="white", fontweight="bold"
    )

    # Reloj de tiempo
    time_txt = ax.text(100, 72, "0.0s", ha='right', fontsize=14, color='white', fontweight='bold')

    # 4. LOOP DE ANIMACIÓN
    def animate(i):
        frame = df_ball.iloc[i]['Frame']
        time_s = df_ball.iloc[i]['Time [s]']
        
        # Posición del Balón
        ball_plot.set_data([df_ball.iloc[i]['x']], [df_ball.iloc[i]['y']])
        
        # Posiciones Jugadores
        h_frame = df_home_long.loc[df_home_long.Frame == frame]
        a_frame = df_away_long.loc[df_away_long.Frame == frame]
        
        home_plot.set_data(h_frame['x'].values, h_frame['y'].values)
        away_plot.set_data(a_frame['x'].values, a_frame['y'].values)
        
        # Actualizar Reloj (Relativo al inicio de la fase)
        rel_time = time_s - df_ball.iloc[0]['Time [s]']
        time_txt.set_text(f"{rel_time:.1f}s")
        
        if show_names:
            # Ocultar todos y resetear
            for t in home_texts + away_texts: t.set_visible(False)
            
            for idx, (_, row) in enumerate(h_frame.iterrows()):
                if idx < len(home_texts):
                    home_texts[idx].set_position((row['x'], row['y'] + 1.8))
                    home_texts[idx].set_text(row['player'])
                    home_texts[idx].set_visible(True)
            
            for idx, (_, row) in enumerate(a_frame.iterrows()):
                if idx < len(away_texts):
                    away_texts[idx].set_position((row['x'], row['y'] + 1.8))
                    away_texts[idx].set_text(row['player'])
                    away_texts[idx].set_visible(True)
            
            return [ball_plot, home_plot, away_plot, time_txt] + home_texts + away_texts
        
        return ball_plot, home_plot, away_plot, time_txt

    # Desactivamos blit si hay texto para evitar artefactos visuales
    anim = animation.FuncAnimation(
        fig, animate, frames=len(df_ball), interval=40, blit=not show_names
    )

    plt.close(fig)
    return HTML(anim.to_jshtml())
```

```python
def generate_executive_report(phases_df, team_name):
    """
    Sintetiza todos los insights tácticos en un informe de rendimiento.
    """
    # 1. Filtramos las jugadas más exitosas (GOL o EPV alto)
    success_phases = phases_df[
        (phases_df['Team'] == team_name) & 
        (phases_df['epv_impact'] > 0.01)
    ].sort_values(by='epv_impact', ascending=False).head(5)
    
    # 2. Filtramos las jugadas de mayor pérdida de valor (Errores críticos)
    error_phases = phases_df[
        (phases_df['Team'] == team_name) & 
        (phases_df['epv_impact'] < -0.01)
    ].sort_values(by='epv_impact', ascending=True).head(5)

    insights_text = "JUGADAS DESTACADAS:\n" + "\n".join(success_phases['Tactical_Insight'].tolist())
    insights_text += "\n\nERRORES A CORREGIR:\n" + "\n".join(error_phases['Tactical_Insight'].tolist())

    prompt = f"""Eres el Analista Jefe. Basado en estos micro-análisis del partido, 
    redacta un Informe Ejecutivo para el entrenador sobre el equipo {team_name}.
    
    {insights_text}
    
    INSTRUCCIONES:
    - Resume el estilo de juego observado.
    - Identifica el principal problema en las pérdidas de balón.
    - Destaca la mayor virtud en la progresión.
    - Usa un tono de alta dirección deportiva.
    """
    
    return generate_with_api(prompt, max_new_tokens=1000)

# Ejemplo:
# report = generate_executive_report(phases_df, "Away")
# print(report)
```

```python
import ipywidgets as widgets
import numpy as np
import pandas as pd
from IPython.display import clear_output, display


# --- 1. FUNCIÓN DE BÚSQUEDA HÍBRIDA ---
def search_pro_tactical(query, min_epv=-1.0, k=5):
    """
    Busca jugadas por significado y filtra por calidad táctica.
    """
    # Búsqueda vectorial
    query_vector = model_embed.encode([query]).astype('float32')
    distances, indices = index.search(query_vector, k * 2) # Buscamos el doble para filtrar
    
    candidates = phases_df.iloc[indices[0]].copy()
    
    # Filtro por KPI de peligrosidad
    results = candidates[candidates['epv_impact'] >= min_epv].head(k)
    
    return results

# --- 2. COMPONENTES DE LA INTERFAZ ---
# Estética
style = {'description_width': 'initial'}
layout_btn = widgets.Layout(width='30%', height='40px')

# Widgets de entrada
search_input = widgets.Text(
    placeholder='Ej: Salida de balón superando líneas por el centro...',
    description='🔍 Consulta Táctica:',
    style=style, layout=widgets.Layout(width='70%')
)
epv_slider = widgets.FloatSlider(
    value=-0.01, min=-0.05, max=0.05, step=0.005,
    description='Min. EPV Impact:',
    style=style, layout=widgets.Layout(width='70%')
)
search_button = widgets.Button(description="BUSCAR JUGADAS", button_style='primary', layout=layout_btn)

# Área de resultados
output_results = widgets.Output()
output_action = widgets.Output()

# Almacén de resultados actuales
current_search_results = None

# --- 3. LÓGICA DE ACCIONES ---

def generate_tactical_summary(b):
    """Opción 1: Generar resumen de las jugadas encontradas"""
    global current_search_results
    with output_action:
        clear_output()
        if current_search_results is None: return
        
        print("🤖 Analizando patrones en las jugadas encontradas...")
        
        # Construimos el contexto para el LLM
        context_text = ""
        for i, (idx, row) in enumerate(current_search_results.iterrows()):
            context_text += f"\nJugada {i+1} (ID {row['Phase_ID']}): {row['Tactical_Insight']}"
            
        prompt = f"""Actúa como el Analista Jefe del Barça. 
        He encontrado estas {len(current_search_results)} jugadas para la consulta: '{search_input.value}'.
        
        DATASET DE JUGADAS:
        {context_text}
        
        TAREA:
        1. Resume el patrón táctico común (¿cómo se progresa?, ¿quién destaca?).
        2. Indica si la ejecución fue eficiente según el EPV.
        3. Da una conclusión de 3 líneas sobre el rendimiento del equipo en este aspecto.
        """
        
        summary = generate_with_api(prompt, max_new_tokens=600, temperature=0.3)
        print(f"\n📝 RESUMEN ESTRATÉGICO:\n{summary}")

def play_selected_video(change):
    """Opción 2: Reproducir el video de la jugada elegida"""
    if not change['new']: return
    phase_id = change['new']
    
    with output_action:
        clear_output()
        print(f"🎬 Generando animación Dark Mode de la Fase {phase_id}...")
        
        # Recuperamos la fila de la fase
        fase = phases_df[phases_df['Phase_ID'] == phase_id].iloc[0]
        
        # Invocamos tu función de tracking (asegúrate de que los tracking_dfs estén cargados)
        anim = plot_pro_tracking_dark(
            tracking_home, tracking_away, tracking_ball, 
            start_frame=fase['Start_Frame'], 
            end_frame=fase['End_Frame'],
            home_team_name="Home", 
            away_team_name="Away"
        )
        display(anim)

# --- 4. CONTROLADOR PRINCIPAL ---

def on_search_clicked(b):
    global current_search_results
    with output_results:
        clear_output()
        with output_action: clear_output()
        
        print("⏳ Buscando en la base de datos vectorial...")
        results = search_pro_tactical(search_input.value, min_epv=epv_slider.value)
        current_search_results = results
        
        if results.empty:
            print("❌ No se encontraron jugadas que cumplan con el criterio de EPV.")
            return

        print(f"✅ Se han encontrado {len(results)} jugadas relevantes:\n")
        
        # Mostrar lista de resultados
        for i, (idx, row) in enumerate(results.iterrows()):
            color = "🟢" if row['epv_impact'] > 0 else "🔴"
            print(f"{i+1}. {color} ID {row['Phase_ID']} | EPV: {row['epv_impact']:.4f} | {row['Outcome']}")
            print(f"   Insight: {row['Tactical_Insight'][:160]}...\n")

        # Configurar botones de acción
        btn_sum = widgets.Button(description="Generar Resumen Grupal", button_style='success', layout=layout_btn)
        btn_sum.on_click(generate_tactical_summary)
        
        video_selector = widgets.Dropdown(
            options=[(f"Analizar Video ID {r['Phase_ID']}", r['Phase_ID']) for _, r in results.iterrows()],
            description='🎥 Ver Jugada:',
            style=style, layout=widgets.Layout(width='40%')
        )
        video_selector.observe(play_selected_video, names='value')
        
        display(widgets.HBox([btn_sum, video_selector]))

search_button.on_click(on_search_clicked)

# --- 5. LANZAR DASHBOARD ---
print("⚽ SISTEMA RAG TÁCTICO INICIADO")
display(widgets.VBox([
    widgets.HTML("<h2>Chatbot Analista RAG - FC Barcelona</h2>"),
    search_input, 
    epv_slider, 
    search_button,
    output_results,
    output_action
]))
```
